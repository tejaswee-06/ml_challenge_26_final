```python
import sqlite3
from pathlib import Path

S2 = Path("/kaggle/working/amazon_ml_solution/index_train/s2_full.sqlite")

print("=" * 70)
print("S2 DATABASE CONTENT VERIFICATION")
print("=" * 70)

con = sqlite3.connect(str(S2))
cur = con.cursor()

# 1. List tables
cur.execute("""
    SELECT name
    FROM sqlite_master
    WHERE type='table'
    ORDER BY name
""")

tables = [r[0] for r in cur.fetchall()]

print("\nTables:")
for t in tables:
    print(" ", t)

# 2. Show schema
print("\nSchema:")
for t in tables:
    cur.execute(f"PRAGMA table_info('{t}')")
    cols = cur.fetchall()
    print(f"\n[{t}]")
    for c in cols:
        print(" ", c)

con.close()
```

```python
import os
from pathlib import Path

print("=" * 80)
print("SEARCHING FOR ANY REMAINING S2 ARTIFACTS")
print("=" * 80)

roots = [
    Path("/kaggle/working"),
    Path("/tmp"),
]

patterns = [
    "*s2_full*",
    "*S2*",
]

for root in roots:
    print(f"\n--- {root} ---")
    try:
        matches = []
        for pattern in patterns:
            matches.extend(root.rglob(pattern))

        seen = set()
        for p in sorted(matches):
            if str(p) in seen:
                continue
            seen.add(str(p))

            try:
                size = p.stat().st_size / (1024**3)
                print(f"{p}    {size:.6f} GB")
            except Exception as e:
                print(f"{p}    [stat failed: {e}]")

    except Exception as e:
        print("SEARCH ERROR:", e)
```

```python
import gc
import sqlite3

print("=" * 80)
print("INSPECTING LIVE S2 SQLITE CONNECTION")
print("=" * 80)

connections = []

for obj in gc.get_objects():
    try:
        if isinstance(obj, sqlite3.Connection):
            connections.append(obj)
    except Exception:
        pass

print("SQLite connections:", len(connections))

# Connection 2 from your previous output
s2_conn = connections[2]

print("\nDatabase:")
print(s2_conn.execute("PRAGMA database_list;").fetchall())

print("\nJournal mode:")
print(s2_conn.execute("PRAGMA journal_mode;").fetchall())

print("\nPage count:")
print(s2_conn.execute("PRAGMA page_count;").fetchall())

print("\nPage size:")
print(s2_conn.execute("PRAGMA page_size;").fetchall())

print("\nIn transaction:")
print(s2_conn.in_transaction)

print("\nSQLite objects:")
tables = s2_conn.execute("""
    SELECT name, type
    FROM sqlite_master
    ORDER BY name;
""").fetchall()

for row in tables:
    print(row)

print("\nSQLite schema:")
schema = s2_conn.execute("""
    SELECT name, type, sql
    FROM sqlite_master
    ORDER BY name;
""").fetchall()

for row in schema:
    print(row)
```
